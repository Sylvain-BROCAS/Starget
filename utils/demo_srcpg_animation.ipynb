```python
import sys
sys.path.append('../utils')
from SRCPG import animate_srcpg_standalone
from IPython.display import HTML
import matplotlib.pyplot as plt
```

## Exemple 1: Configuration simple

- Soleil: 20 dents
- Couronne fixe: 60 dents  
- Couronne sortie: 80 dents
- 3 planètes, module 2mm

```python
# Créer l'animation
anim1 = animate_srcpg_standalone(
    sun_teeth=20,           # Soleil: 20 dents
    fixed_ring_teeth=60,    # Couronne fixe: 60 dents
    output_ring_teeth=80,   # Couronne sortie: 80 dents
    num_planets=3,          # 3 planètes
    module=2.0,             # Module 2mm
    num_frames=120,         # 120 frames pour l'animation
    interval=50             # 50ms entre frames
)

if anim1 is not None:
    print("✅ Animation créée avec succès!")
    HTML(anim1.to_jshtml())
else:
    print("❌ Configuration invalide")
```

## Comment ça marche

### Éléments visualisés:
- 🔴 **Soleil (rouge)**: Entrée qui tourne rapidement
- 🔵 **Couronne fixe (cyan)**: Immobile, bloque les planètes
- 🔵 **Couronne sortie (bleu, pointillée)**: Split ring qui tourne lentement
- 🟡 **Planètes (jaune)**: Orbitent autour du soleil et engrènent avec les couronnes
- ⚫ **Marqueurs**: Indiquent les positions angulaires

### Calculs physiques:
- **Ratio Willis**: $i = \frac{1 + Z_F/Z_S}{1 - Z_F/Z_O}$
- **Vitesses angulaires**: Calculées selon la mécanique des engrenages
- **Positions orbitales**: Les planètes suivent des trajectoires circulaires
