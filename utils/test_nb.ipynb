<section>
<h2>Introduction</h2>

<p>
Ce document présente le principe de fonctionnement du
<strong>Split Ring Compound Planetary Gear (SRCPG)</strong>,
un train d’engrenages compact permettant d’obtenir
des rapports de réduction très élevés (<strong>100:1 à 400:1</strong>)
dans un volume réduit.
</p>

<p>
Le SRCPG combine :
</p>

<ul>
  <li>un <strong>train épicycloïdal classique</strong>,</li>
  <li>un <strong>effet différentiel</strong> basé sur une faible différence
      de dents entre deux couronnes.</li>
</ul>

<div class="info-box">
🎯 Objectif principal :
obtenir un réducteur léger, précis et imprimable en 3D,
adapté aux montures équatoriales goto ultra‑portables.
</div>
</section>

---

<section>
<h2>Principe de fonctionnement</h2>

<div class="image-box">
  
  <div class="caption">
    Source : Juan GG’s blog — Split Ring Compound Epicyclic Gear
  </div>
</div>

<p>
Le SRCPG repose sur deux étages épicycloïdaux couplés via
des planètes composées. Une faible différence de nombre
de dents entre les deux couronnes génère un mouvement
relatif très lent, d’où un fort ratio de réduction.
</p>
</section>

---

<section>
<h2>Équations de dimensionnement</h2>

<p>
Les équations suivantes sont inspirées des travaux de
David Hartkop et permettent de calculer un SRCPG valide
à partir d’un ensemble réduit de paramètres.
</p>

<table>
<thead>
<tr>
  <th>Élément</th>
  <th>Formule</th>
  <th>Raison / intuition</th>
</tr>
</thead>
<tbody>

<tr>
  <td>Planète étage 1</td>
  <td><code>Zp,f = (Zf − Zs) / 2</code></td>
  <td>Condition d’entraxe en engrènement interne</td>
</tr>

<tr>
  <td>Lien des planètes</td>
  <td><code>Zp,o = Zp,f · n</code></td>
  <td>Liberté supplémentaire sur le ratio global</td>
</tr>

<tr>
  <td>Couronne étage 2</td>
  <td><code>Zo = n·Zf + P</code></td>
  <td>Petit écart de dents → effet différentiel</td>
</tr>

<tr>
  <td>Ratio final</td>
  <td><code>Rf = R1 · R2</code></td>
  <td>Multiplication des deux effets cinématiques</td>
</tr>

</tbody>
</table>
</section>


<section>
<h2>Conditions de validité</h2>

<table>
<thead>
<tr>
  <th>Condition</th>
  <th>Pourquoi</th>
</tr>
</thead>
<tbody>

<tr>
  <td><code>(Zf − Zs)</code> pair</td>
  <td>Garantit un nombre entier de dents de planète</td>
</tr>

<tr>
  <td><code>Z % P == 0</code></td>
  <td>Répartition angulaire uniforme des planètes</td>
</tr>

<tr>
  <td><code>Z ≥ 17</code></td>
  <td>Évite la sous‑denture et fragilité des dents</td>
</tr>

</tbody>
</table>

<div class="warning-box">
⚠️ Une configuration valide mathématiquement
peut rester non fabricable sans vérification
géométrique et mécanique.
</div>
</section>

<section>
<h2>Assemblage et fabrication</h2>

<h3>Ordre d’assemblage typique</h3>

<ol>
  <li>Montage des roulements principaux</li>
  <li>Fixation du Soleil sur l’arbre moteur</li>
  <li>Montage des planètes sur le porte‑satellites</li>
  <li>Insertion dans la couronne fixe</li>
  <li>Ajout de la couronne de sortie et fermeture</li>
</ol>

<h3>Jeux recommandés (FDM)</h3>

<ul>
  <li>Jeu dents : 0,15 à 0,20 mm</li>
  <li>Jeu radial planètes : 0,20 à 0,30 mm</li>
  <li>Jeu axial : 0,15 à 0,30 mm</li>
</ul>

<div class="tip-box">
💡 Toujours tester le mécanisme à la main
avant motorisation.
</div>
